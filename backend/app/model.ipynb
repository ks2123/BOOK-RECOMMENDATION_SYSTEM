```python
pivot_df=book_with_rating.copy()
```

```python
pivot_df.columns
```

```python
pt=pivot_df.pivot_table(index="Book-Title",columns='User-ID',values='Book-Rating')
```

```python
pt.fillna(0,inplace=True)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cosine_similarity=cosine_similarity(pt)
```

```python
cosine_similarity.shape
```

```python
import numpy as np

def recommend(book_name):
    # Fetch index of the input book
    index = np.where(pt.index == book_name)[0][0]
    
    # Sort books based on highest similarity score (excluding the book itself at index 0)
    similar_items = sorted(list(enumerate(cosine_similarity[index])), key=lambda x: x[1], reverse=True)[1:6]
    
    # Print top 5 book recommendations
    print(f"\n--- Recommendations for '{book_name}' ---")
    for i in similar_items:
        print(f"📖 {pt.index[i[0]]}")

# Example Run (Kisi bhi valid book title ke saath test karo):
recommend('1984')
```

```python
import numpy as np

def recommend(book_name):
    # 1. Book ka exact index fetch karo
    index = np.where(pt.index == book_name)[0][0]
    
    # 2. Similarity scores ko pair (index, score) mein convert karke SCORE ke basis par desc sort karo
    scores = list(enumerate(cosine_similarity[index]))
    sorted_scores = sorted(scores, key=lambda x: x[1], reverse=True)
    
    # 3. Top 5 items pick karo (index 0 self book hoti hai, so index 1 to 6)
    top_recommendations = sorted_scores[1:6]
    
    print(f"--- Real Similarity Recommendations for '{book_name}' ---")
    for i in top_recommendations:
        book_idx = i[0]
        similarity_val = i[1]
        print(f"📖 {pt.index[book_idx]} (Score: {round(similarity_val, 3)})")

# Test Karo:
recommend('1984')
```
